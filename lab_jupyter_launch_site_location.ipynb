Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

*TODO:* Create a new column in `spacex_df` dataframe called `marker_color` to store the marker colors based on the `class` value


```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
import folium
from folium.plugins import MarkerCluster

# 1. Initialize the MarkerCluster object as instructed
marker_cluster = MarkerCluster()

# 2. TODO: Create the marker_color column in spacex_df
# We apply a function that returns 'green' if class is 1, else 'red'
spacex_df['marker_color'] = spacex_df['class'].apply(lambda x: 'green' if x == 1 else 'red')

# Let's preview the dataframe to make sure the column was added correctly
spacex_df[['Launch Site', 'class', 'marker_color']].head()
```

*TODO:* For each launch result in `spacex_df` data frame, add a `folium.Marker` to `marker_cluster`


```python
# 1. Add the marker_cluster container to the current site_map first
site_map.add_child(marker_cluster)

# 2. Loop through each row in the spacex_df dataframe
for index, record in spacex_df.iterrows():
    # Extract the coordinate for the current launch record
    coordinate = [record['Lat'], record['Long']]
    
    # TODO: Create a Marker object with custom icon colors based on launch success/failure
    marker = folium.Marker(
        location=coordinate,
        # Popups are great for showing details when clicking a specific pin
        popup=f"Site: {record['Launch Site']}\nStatus: {'Success' if record['class'] == 1 else 'Fail'}",
        icon=folium.Icon(color='white', icon_color=record['marker_color'])
    )
    
    # TODO: Add the marker to the marker_cluster container
    marker_cluster.add_child(marker)

# Display the final updated map
site_map
```

Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
# 1. Define the launch site coordinates
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# 2. Define the coastline coordinates (obtained via MousePosition on your map)
coastline_lat = 28.56367
coastline_lon = -80.56796  # Adjusted slightly to hit the actual beach line east of SLC-40

# 3. Calculate the distance using our function
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distance to closest coastline: {distance_coastline:.2f} km")
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )
# 1. Define the coordinates for the marker
coastline_coordinate = [coastline_lat, coastline_lon]

# 2. Create the distance marker on the closest coastline point
distance_marker = folium.Marker(
    coastline_coordinate,
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        # Using your template formatting to inject the distance value dynamically
        html='<div style="font-size: 12px; color:#d35400; width: 150px;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
    )
)

# 3. Add the distance marker to your site map
site_map.add_child(distance_marker)

# Display the updated map
site_map
```

```python
# Create a `folium.PolyLine` object using the coastline coordinates and launch site coordinate
# lines=folium.PolyLine(locations=coordinates, weight=1)
# 1. Combine the launch site and coastline coordinates into a list
coordinates = [[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]]

# 2. TODO: Create the folium.PolyLine object
# Setting a weight of 1 creates a clean, precise line, and color='blue' makes it stand out
lines = folium.PolyLine(locations=coordinates, weight=1, color='blue')

# 3. Add the line to the map
site_map.add_child(lines)

# Display the map with your new connecting line
site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
# Launch Site: CCAFS SLC-40
launch_site_lat = 28.563197
launch_site_lon = -80.576820
launch_coordinate = [launch_site_lat, launch_site_lon]

# Dictionary of closest points of interest (POIs) determined by regional geography
closest_pois = {
    'Nearest Railway': {
        'lat': 28.57211, 
        'lon': -80.58528, 
        'color': 'green'
    },
    'Nearest Highway (NASA Pkwy)': {
        'lat': 28.52562, 
        'lon': -80.64835, 
        'color': 'blue'
    },
    'Nearest City (Titusville)': {
        'lat': 28.61150, 
        'lon': -80.80770, 
        'color': 'purple'
    }
}
# Loop through each point of interest automatically
for poi_name, poi_data in closest_pois.items():
    poi_coordinate = [poi_data['lat'], poi_data['lon']]
    
    # 1. Automatically calculate the distance
    distance = calculate_distance(
        launch_site_lat, 
        launch_site_lon, 
        poi_data['lat'], 
        poi_data['lon']
    )
    
    # 2. Create a clean text label showing the POI name and calculated distance
    label_text = f"{poi_name}: {distance:.2f} KM"
    
    distance_marker = folium.Marker(
        poi_coordinate,
        icon=DivIcon(
            icon_size=(250, 20),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12px; color:{poi_data["color"]}; width: 250px;"><b>{label_text}</b></div>',
        )
    )
    site_map.add_child(distance_marker)
    
    # 3. Draw the PolyLine connecting the launch site to this POI
    line_coordinates = [launch_coordinate, poi_coordinate]
    line = folium.PolyLine(
        locations=line_coordinates, 
        weight=2, 
        color=poi_data['color'],
        opacity=0.8
    )
    site_map.add_child(line)

# Display the map with all infrastructure lines rendered dynamically
site_map
```
